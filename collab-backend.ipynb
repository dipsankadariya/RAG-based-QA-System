````python
import os
import subprocess

REPO_URL = "https://github.com/dipsankadariya/RAG-based-QA-System.git"
BRANCH = "dipsan-forum"
REPO_DIR = "/content/RAG-based-QA-System"
BACKEND_DIR = os.path.join(REPO_DIR, "nepali-legal-qa", "backend")

if not os.path.exists(REPO_DIR):
    print("Cloning repo...")
    subprocess.run(["git", "clone", "-b", BRANCH, REPO_URL, REPO_DIR], check=True)
else:
    print("Repo already cloned. Updating...")
    os.chdir(REPO_DIR)
    subprocess.run(["git", "fetch", "origin"], check=True)
    subprocess.run(["git", "checkout", BRANCH], check=True)
    subprocess.run(["git", "pull", "origin", BRANCH], check=True)

os.chdir(BACKEND_DIR)
print(f"Working directory: {os.getcwd()}")

print("Installing dependencies...")
subprocess.run(["pip", "install", "-q", "-r", "requirements.txt"], check=True)
subprocess.run(["pip", "install", "-q", "pyngrok"], check=True)
print("Done.")
````

## Cell 4.2 — (Optional) Agent keys (PageIndex + Gemini)

If you want **Agent mode** to work, add `PAGEINDEX_API_KEY` and `GOOGLE_API_KEY` in Colab Secrets and run the next cell.
If you skip this, HyDE RAG will still work.

````python
import os
from google.colab import userdata

# These are required only for Agent mode
try:
    pageindex_api_key = userdata.get("PAGEINDEX_API_KEY")
except Exception:
    pageindex_api_key = None

try:
    google_api_key = userdata.get("GOOGLE_API_KEY")
except Exception:
    google_api_key = None

try:
    agent_model = userdata.get("AGENT_MODEL")
except Exception:
    agent_model = None

try:
    pageindex_mcp_url = userdata.get("PAGEINDEX_MCP_URL")
except Exception:
    pageindex_mcp_url = None

if not pageindex_api_key or not google_api_key:
    print("Agent mode not configured (missing PAGEINDEX_API_KEY and/or GOOGLE_API_KEY).")
else:
    os.environ["PAGEINDEX_API_KEY"] = pageindex_api_key
    os.environ["GOOGLE_API_KEY"] = google_api_key
    os.environ["AGENT_MODEL"] = agent_model or "gemini-3.1-flash-lite"
    if pageindex_mcp_url:
        os.environ["PAGEINDEX_MCP_URL"] = pageindex_mcp_url
    print("Agent keys set")
````


**API Response Fields:**
- `mode` — `hyde` or `agent`
- `answer` — single best answer
- `answer_in_english` — optional translation (may be null)
- `hyde_passage` — only for `hyde` mode
- `retrieved_docs` — only for `hyde` mode
